```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import warnings

# Suppress visual clutter from standard library warnings
warnings.filterwarnings('ignore')

# Set strict, professional visual standards for DAAD-level reporting
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams.update({'figure.figsize': (10, 6), 'figure.dpi': 120})
```

```python
# The notebook is inside 'notebooks/', so we step up one directory to access 'artifacts/'
data_path = os.path.join("..", "artifacts", "data.csv")

try:
    df = pd.read_csv(data_path)
    print(f"Dataset Shape: {df.shape}")
    display(df.head())
    print("\nDataset Schema and Memory Usage:")
    df.info()
except FileNotFoundError:
    print(f"CRITICAL ERROR: Artifact not found at {data_path}. Pipeline failure.")
```

```python
# Cell 3: Label Distribution and Imbalance Quantification
plt.figure(figsize=(8, 5))
ax = sns.countplot(
    data=df, 
    x='label', 
    order=df['label'].value_counts().index, 
    palette='viridis'
)
plt.title('Distribution of Civic Grievance Categories', fontsize=14, fontweight='bold')
plt.xlabel('Grievance Category', fontsize=12)
plt.ylabel('Absolute Frequency', fontsize=12)

# Annotate bars with exact percentages
total = len(df)
for p in ax.patches:
    percentage = f'{100 * p.get_height() / total:.1f}%'
    x = p.get_x() + p.get_width() / 2 - 0.15
    y = p.get_height() + 0.05
    ax.annotate(percentage, (x, y), size=10, fontweight='bold')

plt.show()

print("\nMathematical Class Distribution (Percentages):")
display(df['label'].value_counts(normalize=True) * 100)
```

```python
%matplotlib inline

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x='label',
    order=df['label'].value_counts().index,
    palette='viridis'
)

plt.title('Distribution of Civic Grievance Categories')
plt.xlabel('Grievance Category')
plt.ylabel('Absolute Frequency')

plt.show()
```

```python
# Cell 4: Text Length Distribution Analysis

# Engineer length features
df['char_count'] = df['text'].apply(len)
df['word_count'] = df['text'].apply(lambda x: len(str(x).split()))

# Plot distributions
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Character Count Histogram
sns.histplot(df['char_count'], bins=20, kde=True, ax=axes[0], color='blue')
axes[0].set_title('Character Count Distribution', fontweight='bold')
axes[0].set_xlabel('Number of Characters')
axes[0].set_ylabel('Frequency')

# Word Count Histogram
sns.histplot(df['word_count'], bins=20, kde=True, ax=axes[1], color='orange')
axes[1].set_title('Word Count Distribution', fontweight='bold')
axes[1].set_xlabel('Number of Words')
axes[1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()

# Calculate and display the 95th percentile
word_95th = df['word_count'].quantile(0.95)

print(f"\n95th Percentile Word Count: {word_95th:.1f} words")
print("This value will inform our Transformer tokenization strategy.")
```
